# Set Data Structure

A set is an unordered collection of unique values. Sets are mutable, but they do not preserve order and they do not allow duplicate elements.

## Main set features
- Unique items only
- Unordered collection
- Mutable: you can add and remove values
- Useful for membership tests, removing duplicates, and set operations

```python
numbers = {1, 2, 3, 4}
print(numbers)

# Add a new value
numbers.add(5)
print('After add:', numbers)

# Adding a duplicate does nothing
numbers.add(3)
print('After duplicate add:', numbers)

# Membership test
print('Contains 2?', 2 in numbers)

# Remove values
numbers.remove(4)
print('After remove:', numbers)

# Set operations
other = {3, 5, 7}
print('Union:', numbers | other)
print('Intersection:', numbers & other)
print('Difference:', numbers - other)
```
