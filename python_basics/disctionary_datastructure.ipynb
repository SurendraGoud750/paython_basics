# Dictionary Data Structure

A dictionary stores items as key-value pairs. Keys must be unique and are used to look up values quickly. Dictionaries are mutable, so you can add, update, and remove items after creation.

## Main dictionary methods
- `dict[key]` or `dict.get(key)` to read values
- `dict[key] = value` to add or update an item
- `dict.pop(key)` to remove an item
- `dict.keys()`, `dict.values()`, `dict.items()` to inspect contents
- `'key' in dict` to test membership
- `dict.clear()` to empty the dictionary
- `dict.update(other_dict)` to merge another dictionary

```python
student = {
    'name': 'Surendra',
    'age': 40,
    'score': 40.50,
    'city': 'Mumbai'
}

print(student['name'])
print(student.get('age'))
print('All values:', student)

# Update an existing value
student['age'] = 41
# Add a new key-value pair
student['email'] = 'surendra@example.com'
print('Updated:', student)

print('Keys:', list(student.keys()))
print('Values:', list(student.values()))
print('Items:', list(student.items()))
print('Has name?', 'name' in student)

# Remove a key
student.pop('score')
print('After removing score:', student)
```
